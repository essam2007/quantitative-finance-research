# Alpha Decay Research

**Goal:** Measure decay in strategy performance (e.g. Sharpe ratio) after publication or widespread adoption.

1. Define factor strategy and pre/post publication windows.
2. Backtest pre-publication → Sharpe pre.
3. Backtest post-publication → Sharpe post.
4. Report decay: ratio or difference in Sharpe.

```python
import sys; sys.path.insert(0, '../..')
from quant_builds.alpha_decay import simulate_factor_returns, alpha_decay_report, sharpe
import numpy as np
```

## Synthetic example: one factor, pre vs post

```python
n_pre, n_post = 252 * 5, 252 * 5
pre = simulate_factor_returns(n_pre, sharpe_target=0.6, decay_after=None, seed=42)
post = simulate_factor_returns(n_post, sharpe_target=0.6, decay_after=int(n_post * 0.1), decay_slope=0.4, seed=43)
report = alpha_decay_report(pre, post)
print("Sharpe pre:", report["Sharpe_pre"])
print("Sharpe post:", report["Sharpe_post"])
print("Ratio (post/pre):", report["Sharpe_ratio_post_pre"])
```
